```python
import numpy as np
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score
from sklearn.model_selection import GridSearchCV

from sklearn.ensemble import RandomForestRegressor
```

```python
from datasets.sequences import SequencesDataset
```

```python
SEQUENCE_LENGTH = 5
sequences_dataset = SequencesDataset(
  sequence_length = SEQUENCE_LENGTH,
  split="",
  thermal_root_folder="../../datasets/thermal",
)

total_items = len(sequences_dataset)
print(f"Total Items: {total_items}")
```

```python
X = []
y = []

for index in range(total_items):
  sequence = sequences_dataset.__getitem__(index)[0]
  coefficients = sequences_dataset.__getitem__(index)[1]
  sequence = sequence.flatten()
  X.append(sequence)
  y.append(coefficients)

X = np.array(X)
y = np.array(y)

print(f"X Shape: {X.shape}")
print(f"y Shape: {y.shape}")
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.2, random_state = 0)
print(f"X_train Shape: {X_train.shape}")
print(f"y_train Shape: {y_train.shape}")
print(f"X_test Shape: {X_test.shape}")
print(f"y_test Shape: {y_test.shape}")
```

```python
rf_regressor = RandomForestRegressor(n_estimators=100, random_state=0)
rf_regressor.fit(X_train, y_train)
```

```python
y_pred_train = rf_regressor.predict(X_train)
r2_score(y_train, y_pred_train)
```

```python
y_pred = rf_regressor.predict(X_test)
r2_score(y_test, y_pred)
```
