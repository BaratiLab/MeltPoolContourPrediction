```python
def find_contour_extremum(extremum="max"):
    """
    Determines min or max extremum value for either y or x values
    respectively with an initial read of all the contour data.
    """
    contour_extrema_values = []

    for file in glob.glob(f"{CONTOURS_FOLDER}/*.csv"):
        micron_values = np.loadtxt(file, skiprows = 1, delimiter = ',')

        if (extremum == "max"):
            # Looks through x values to find the maximum melt pool width.
            contour_extrema_values.append(np.max(micron_values[:, 0]))

        if (extremum == "min"):
            # Looks through y values to find the maximum melt pool depth.
            contour_extrema_values.append(np.min(micron_values[:, 1]))
    print(contour_extrema_values)
    if (extremum == "max"):
        return max(contour_extrema_values)

    if (extremum == "min"):
        return min(contour_extrema_values)
```

```python
def normalize_and_define_contour_parameters(min_y, max_x):
    """
    Normalizes and keeps track of the corresponding parameter values for
    each contour.
    """
    contour_normalized_values = []
    contour_parameters = []

    for file in glob.glob(f"{CONTOURS_FOLDER}/*.csv"):
        # Normalizes the contour values.
        micron_values = np.loadtxt(file, skiprows = 1, delimiter = ',')

        normalized_x = micron_values[:, 0] / max_x
        normalized_y = micron_values[:, 1] / min_y

        # contour_normalized_values.append(binary_area_plot)
        contour_normalized_values.append([normalized_x, normalized_y])

        # Creates a seperate list to keep track of the contour parameters.

        # Removes directory info (This is made on windows)
        # filename = file.split("\\")[-1]

        # Removes directory info (This is made on macos)
        filename = file.split("/")[-1]

        # Removes extension name
        filename = filename.split(".csv")[0]

        # Splits by "_" delimiter
        filename_info = filename.split("_")

        bead_number = int(filename_info[3])
        set_number = int(filename_info[0].split("Set")[1]) 

        power_velocity = get_power_velocity(set_number, bead_number)

        contour_parameters.append({
            "set": set_number,
            "series": int(filename_info[1]),
            "bead": bead_number,
            "power": power_velocity[0],
            "velocity": power_velocity[1] 
        })

    return contour_normalized_values, contour_parameters, micron_values
```

```python
from tqdm import tqdm
import json
import skimage as ski
from skimage.measure import block_reduce
def convert_contour_to_area(contour_normalized_values, contour_parameters, img_size = 64, method = 'max', series = False):
    """
    Converts the contour values to a binary area plot.
    """
    contour_parameter_map = dict()

    for index, contour_parameter in tqdm(enumerate(contour_parameters)):
        contour_normalized_value = contour_normalized_values[index]
        new_contour_x = np.append(contour_normalized_value[0], contour_normalized_value[0][0])
        new_contour_y = np.append(-contour_normalized_value[1], -contour_normalized_value[1][-1])
        image_indices_x = new_contour_x*64
        image_indices_y = 64 + new_contour_y*64 - 1
        image_shape = (1024,1024)
        image = np.zeros(image_shape, dtype=bool)
        mask = ski.draw.polygon2mask(image_shape, np.array([16*image_indices_x, 16*(62-image_indices_y)]).T)
        mask = mask.T
        block_reduce_factor = int(np.round(1024/img_size))
        if method == 'max':
            downsampled_mask = block_reduce(mask, (block_reduce_factor, block_reduce_factor), np.max)
        elif method == 'mean':
            downsampled_mask = block_reduce(mask, (block_reduce_factor, block_reduce_factor), np.mean)
        else:
            raise ValueError(f"Method {method} not supported")
        power = contour_parameter["power"]
        velocity = contour_parameter["velocity"]
        if not series:
            contour_parameter_map[f"{power}_{velocity}"] = list(downsampled_mask.flatten().astype(float))
        else:
            if f"{power}_{velocity}" not in contour_parameter_map:
                contour_parameter_map[f"{power}_{velocity}"] = []
            else:
                contour_parameter_map[f"{power}_{velocity}"].append(list(downsampled_mask.flatten().astype(float)))
    return contour_parameter_map
# convert_contour_to_area(contour_normalized_values[90])
root_dir = os.path.abspath(os.path.join(os.getcwd(), '../../../../'))

img_size = 128
method = 'max'
for method in ['max', 'mean']:
    contour_parameter_map = convert_contour_to_area(contour_normalized_values, contour_parameters, img_size = img_size, method = method, series = True)
    with open(os.path.join(root_dir, f"./datasets/input_label_map/{img_size}_enlarged_{method}_contour_area_series.json"), "w") as fp:
        json.dump(contour_parameter_map , fp, indent = 4)
```
