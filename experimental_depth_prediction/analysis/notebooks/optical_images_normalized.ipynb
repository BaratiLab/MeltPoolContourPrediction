# Optical Images Normalized
Normalizes optical images from microns to values between 0 and 1.

1 being the minimum (Y-axis) or maximum (X-axis) value for each axis.

```python
import csv
import glob
import matplotlib.pyplot as plt
import numpy as np

fields = ["X", "Y"]
directory = "../../data/Optical_Images_Etched/Normalized"
```

```python
# All Sets from microns folder
csv_files = glob.glob('../../data/Optical_Images_Etched/Micron/*Set*/*csv')

max_x_values = []
min_y_values = []

for file in csv_files:
    micron_values = np.loadtxt(file, skiprows = 1, delimiter = ',')
    max_x_values.append(np.max(micron_values[:, 0]))
    min_y_values.append(np.min(micron_values[:, 1]))

absolute_max_x = max(max_x_values)
absolute_min_y = min(min_y_values)
print(f"Absolute max x: {absolute_max_x}")
print(f"Absolute min y: {absolute_min_y}")

for file in csv_files:
    # Make sure the file names are correct here.
    filename = f"{file[50:-4]}.csv"
    print(f"Saving as `{filename}`")
    micron_values = np.loadtxt(file, skiprows = 1, delimiter = ',')

    normalized_x = micron_values[:, 0] / absolute_max_x
    normalized_y = micron_values[:, 1] / absolute_min_y

    # Save Values
    normalized_values = np.array([normalized_x, normalized_y]).T

    with open(f"{directory}/{filename}", "w", newline="") as f:
        write = csv.writer(f)
        write.writerow(fields)
        write.writerows(normalized_values)
```

```python
csv_files = glob.glob('../../data/Optical_Images_Etched/Normalized/*csv')

plt.figure(dpi=150)

for file in csv_files:
    normalized_values = np.loadtxt(file, skiprows = 1, delimiter = ',')
    plt.plot(normalized_values[:,0], normalized_values[:,1], alpha =0.5)
    plt.axis('equal')
    plt.ylabel('Normalized Depth')
    plt.xlabel('Normalized Width')

plt.xlim(0, 1)
plt.ylim(0, 1)
plt.gca().invert_yaxis()
```
