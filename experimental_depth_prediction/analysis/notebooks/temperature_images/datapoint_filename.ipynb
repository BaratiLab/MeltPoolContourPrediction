# Breaks down file name into informative bits 

````python
import glob
````

The files in the dataset are structured with the following syntax

```bash
316SS_Mono_0.3mpersec_50W_20us_Frame1.tif
```

```python
f"316SS_{image_type}_{velocity}mpersec_{power}W_{exposure}us_Frame{frame_number}.tif"
```

````python
# Parse file name into useful bits
image_files = glob.glob("../../../data/1_Temperature_4us_Exposure_Time/*")
image_file = image_files[0]

# Removes directory info (This is made on windows)
filename = image_file.split("\\")[-1]

# Removes extension name
filename = filename.split(".tif")[0]

print(f"Filename: {filename} \n")

# Splits by "_" delimiter
filename_info = filename.split("_")


image_type = filename_info[1]
velocity = float(filename_info[2].split("mpersec")[0])
power = int(filename_info[3].split("W")[0])
exposure = int(filename_info[4].split("us")[0])
frame = int(filename_info[5].split("Frame")[-1])

print("File Name Information")
print(f"Filename Info: {filename_info}")
print(f"Image Type: {image_type}")
print(f"Velocity (m/s): {velocity}")
print(f"Power (W): {power}")
print(f"Exposure (us): {exposure}")
print(f"Frame: {frame}")
````
